```python
import xml.etree.ElementTree as ET

# Select the VV annotation file
vv_annotation = next(
    f for f in annotation_files
    if "-vv-" in f.name.lower()
)

print("VV annotation:")
print(vv_annotation)

# Parse XML
tree = ET.parse(vv_annotation)
root = tree.getroot()

print("\nXML root:")
print(root.tag)

# Search for geolocation grid points
geo_points = root.findall(".//geolocationGridPoint")

print("\nGeolocation grid points found:", len(geo_points))

if geo_points:
    point = geo_points[0]

    print("\nFirst geolocation point:")
    for child in point:
        print(f"  {child.tag.split('}')[-1]} = {child.text}")
```

```python
# Find the geolocation grid point closest to Ennore

TARGET_LAT = 13.2282
TARGET_LON = 80.3633

geo_data = []

for point in geo_points:
    values = {}

    for child in point:
        tag = child.tag.split("}")[-1]
        values[tag] = child.text

    try:
        values["line"] = int(values["line"])
        values["pixel"] = int(values["pixel"])
        values["latitude"] = float(values["latitude"])
        values["longitude"] = float(values["longitude"])
        geo_data.append(values)
    except (KeyError, ValueError):
        continue

print("Valid geolocation points:", len(geo_data))

# Find closest point using simple geographic distance
closest = min(
    geo_data,
    key=lambda p: (
        (p["latitude"] - TARGET_LAT) ** 2 +
        (p["longitude"] - TARGET_LON) ** 2
    )
)

print("\n🎯 Closest geolocation point to Ennore")
print("=" * 50)

print("Target latitude :", TARGET_LAT)
print("Target longitude:", TARGET_LON)

print("\nClosest SAR point:")
print("Line           :", closest["line"])
print("Pixel          :", closest["pixel"])
print("Latitude       :", closest["latitude"])
print("Longitude      :", closest["longitude"])

distance = (
    (closest["latitude"] - TARGET_LAT) ** 2 +
    (closest["longitude"] - TARGET_LON) ** 2
) ** 0.5

print("\nApprox. coordinate difference:", distance)
```

```python
import numpy as np
from scipy.interpolate import LinearNDInterpolator

# Arrays from Sentinel-1 geolocation grid
lats = np.array([p["latitude"] for p in geo_data])
lons = np.array([p["longitude"] for p in geo_data])
lines = np.array([p["line"] for p in geo_data], dtype=float)
pixels = np.array([p["pixel"] for p in geo_data], dtype=float)

# Build interpolation:
# geographic coordinates -> SAR image coordinates
line_interpolator = LinearNDInterpolator(
    np.column_stack((lats, lons)),
    lines
)

pixel_interpolator = LinearNDInterpolator(
    np.column_stack((lats, lons)),
    pixels
)

# Estimate Ennore's SAR coordinates
ennore_line = line_interpolator(TARGET_LAT, TARGET_LON)
ennore_pixel = pixel_interpolator(TARGET_LAT, TARGET_LON)

print("🎯 Interpolated Ennore SAR position")
print("=" * 50)

print("Latitude :", TARGET_LAT)
print("Longitude:", TARGET_LON)

print("\nEstimated image position:")

if np.isnan(ennore_line) or np.isnan(ennore_pixel):
    print("❌ Interpolation failed.")
    print("Target may be outside the geolocation grid.")
else:
    print("Line   :", float(ennore_line))
    print("Pixel  :", float(ennore_pixel))
```

```python
# Crop a local region around Ennore
# We use the VV image first.

center_line = int(round(float(ennore_line)))
center_pixel = int(round(float(ennore_pixel)))

half_size = 2000

line_start = max(0, center_line - half_size)
line_end = min(vv.shape[0], center_line + half_size)

pixel_start = max(0, center_pixel - half_size)
pixel_end = min(vv.shape[1], center_pixel + half_size)

vv_ennore = vv[line_start:line_end, pixel_start:pixel_end]
vh_ennore = vh[line_start:line_end, pixel_start:pixel_end]

print("🎯 Ennore SAR crop")
print("=" * 45)

print("Center line :", center_line)
print("Center pixel:", center_pixel)

print("\nCrop bounds:")
print("Lines :", line_start, "to", line_end)
print("Pixels:", pixel_start, "to", pixel_end)

print("\nCrop dimensions:")
print("VV:", vv_ennore.shape)
print("VH:", vh_ennore.shape)
```

```python
# Display the Ennore VV and VH crop

def robust_limits(image):
    valid = image[image > 0]
    return np.percentile(valid, [2, 98])

vv_min, vv_max = robust_limits(vv_ennore)
vh_min, vh_max = robust_limits(vh_ennore)

plt.figure(figsize=(10, 8))

plt.imshow(
    vv_ennore,
    cmap="gray",
    vmin=vv_min,
    vmax=vv_max
)

plt.scatter(
    center_pixel - pixel_start,
    center_line - line_start,
    marker="+",
    s=150
)

plt.title("Ennore Region — Sentinel-1A VV")
plt.xlabel("Pixel")
plt.ylabel("Line")
plt.colorbar(label="Raw DN")
plt.show()


plt.figure(figsize=(10, 8))

plt.imshow(
    vh_ennore,
    cmap="gray",
    vmin=vh_min,
    vmax=vh_max
)

plt.scatter(
    center_pixel - pixel_start,
    center_line - line_start,
    marker="+",
    s=150
)

plt.title("Ennore Region — Sentinel-1A VH")
plt.xlabel("Pixel")
plt.ylabel("Line")
plt.colorbar(label="Raw DN")
plt.show()
```

```python
from scipy import ndimage
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Convert raw values to relative normalized values
# --------------------------------------------------

vv_float = vv_ennore.astype(np.float32)
vh_float = vh_ennore.astype(np.float32)

# Ignore zero/background
vv_valid = vv_float[vv_float > 0]
vh_valid = vh_float[vh_float > 0]

# Robust normalization using percentiles
vv_low, vv_high = np.percentile(vv_valid, [2, 98])
vh_low, vh_high = np.percentile(vh_valid, [2, 98])

vv_norm = np.clip(
    (vv_float - vv_low) / (vv_high - vv_low),
    0,
    1
)

vh_norm = np.clip(
    (vh_float - vh_low) / (vh_high - vh_low),
    0,
    1
)

# --------------------------------------------------
# 2. Identify ocean pixels
# --------------------------------------------------
# Land is generally much brighter in this scene.
# Use a conservative brightness threshold.

land_threshold = np.percentile(vv_valid, 65)

ocean_mask = vv_float < land_threshold

# Remove background zeros
ocean_mask &= vv_float > 0
ocean_mask &= vh_float > 0

# --------------------------------------------------
# 3. Find unusually dark pixels
# --------------------------------------------------

vv_dark_threshold = np.percentile(vv_valid, 15)
vh_dark_threshold = np.percentile(vh_valid, 15)

vv_dark = vv_float < vv_dark_threshold
vh_dark = vh_float < vh_dark_threshold

# Require both polarizations to be relatively dark
candidate_mask = (
    ocean_mask &
    vv_dark &
    vh_dark
)

# --------------------------------------------------
# 4. Remove tiny isolated objects
# --------------------------------------------------

labeled, num_features = ndimage.label(candidate_mask)

sizes = np.bincount(labeled.ravel())

min_pixels = 100

clean_mask = np.zeros_like(candidate_mask, dtype=bool)

for label_id in range(1, len(sizes)):
    if sizes[label_id] >= min_pixels:
        clean_mask[labeled == label_id] = True

print("🔎 MARIS baseline detection")
print("=" * 45)

print("VV dark threshold:", vv_dark_threshold)
print("VH dark threshold:", vh_dark_threshold)

print("Initial candidate pixels:",
      int(candidate_mask.sum()))

print("Clean candidate pixels:",
      int(clean_mask.sum()))

print("Candidate regions:",
      int((sizes[1:] >= min_pixels).sum()))
```
